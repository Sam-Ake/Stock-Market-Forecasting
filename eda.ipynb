```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# Check possible locations for the stocks folder
possible_paths = [
    "stocks/aapl.us.csv",
    "../stocks/aapl.us.csv",
    "aapl.us.csv",
]

df = None
for path in possible_paths:
  if os.path.exists(path):
    df = pd.read_csv(path)
    print(f"Successfully loaded dataset from: {path}")
    break

if df is None:
  print(
      "❌ Could not find 'aapl.us.csv'. Please check where your extracted"
      " Kaggle files are located!"
  )
else:
  # Basic Cleaning & Indexing
  df["Date"] = pd.to_datetime(df["Date"])
  df = df.sort_values("Date")
  df.set_index("Date", inplace=True)

  ts_data = df["Close"].dropna()

  # Visualize Price Trends and Moving Averages
  plt.figure(figsize=(14, 6))
  plt.plot(ts_data, label="Daily Close Price", color="royalblue", alpha=0.7)
  plt.plot(
      ts_data.rolling(window=30).mean(),
      label="30-Day Moving Average",
      color="darkorange",
      linewidth=2,
  )
  plt.plot(
      ts_data.rolling(window=365).mean(),
      label="365-Day Moving Average",
      color="forestgreen",
      linewidth=2,
  )
  plt.title(
      "AAPL Stock Price & Moving Averages", fontsize=14, fontweight="bold"
  )
  plt.xlabel("Date")
  plt.ylabel("Price (USD)")
  plt.legend()
  plt.grid(True, linestyle="--", alpha=0.5)
  plt.tight_layout()
  plt.show()

  # ACF and PACF Plots
  fig, axes = plt.subplots(1, 2, figsize=(16, 5))
  plot_acf(
      ts_data, lags=40, ax=axes[0], title="Autocorrelation Function (ACF)"
  )
  axes[0].set_xlabel("Lags")
  axes[0].set_ylabel("Correlation")

  plot_pacf(
      ts_data,
      lags=40,
      ax=axes[1],
      title="Partial Autocorrelation Function (PACF)",
      method="ywm",
  )
  axes[1].set_xlabel("Lags")
  axes[1].set_ylabel("Correlation")

  plt.suptitle("Time Series Diagnostic Plots", fontsize=16, y=1.03)
  plt.tight_layout()
  plt.show()
```

```python
import glob
import os
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller

# 1. Load a valid stock CSV from your dataset folders
# Let's find any available stock or ETF CSV automatically from the glob results
csv_files = glob.glob("**/*.csv", recursive=True)
stock_files = [f for f in csv_files if "symbols_valid" not in f]

if not stock_files:
  raise FileNotFoundError(
      "No stock CSV files found! Please check your dataset folder structure."
  )

file_path = stock_files[0]  # Picks the first available stock file
print(f"Loading dataset from: {file_path}")

df = pd.read_csv(file_path)

# 2. Clean and Index by Date
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date")
df.set_index("Date", inplace=True)

# Focus on 'Close' price
ts = df["Close"].dropna()


# 3. Test for Stationarity (Augmented Dickey-Fuller Test)
def test_stationarity(series):
  result = adfuller(series.dropna())
  print(f"ADF Statistic: {result[0]:.4f}")
  print(f"p-value: {result[1]:.4f}")
  if result[1] <= 0.05:
    print(
        "✨ Series is Stationary (Reject H0 - no unit root present). Ready for"
        " modeling!"
    )
  else:
    print(
        "⚠️ Series is Non-Stationary (Fail to reject H0 - unit root present)."
        " Differencing required."
    )


print("--- Original Series Stationarity Test ---")
test_stationarity(ts)

# 4. Apply Differencing to Achieve Stationarity
# First-order differencing
ts_diff = ts.diff().dropna()

print("\n--- Differenced Series Stationarity Test ---")
test_stationarity(ts_diff)

# 5. Create Feature Engineering Dataset (Lag Features)
# We combine the original series and differenced series into a supervised learning frame
data = pd.DataFrame({"Close": ts})

# Create 5 lag features
for i in range(1, 6):
  data[f"lag_{i}"] = data["Close"].shift(i)

# Create rolling mean features for trend capture
data["rolling_mean_7"] = data["Close"].shift(1).rolling(window=7).mean()
data["rolling_mean_30"] = data["Close"].shift(1).rolling(window=30).mean()

# Drop rows with NaN values resulting from lags/rolling windows
data = data.dropna()

print("\n--- Preprocessed Feature Dataset Head ---")
print(data.head())
print(f"\nFinal dataset shape for modeling: {data.shape}")
```

```python
# Keep only data from January 1, 2023 to the most recent available date
df = df.loc['2023-01-01':]

print(f'Filtered dataset date range: {df.index.min().date()} to {df.index.max().date()}')
print(f'Total records remaining: {len(df)}')
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_percentage_error, mean_squared_error
from sklearn.model_selection import TimeSeriesSplit

# 1. Define Features (X) and Target (y)
features = [col for col in data.columns if col != "Close"]
X = data[features]
y = data["Close"]

print(f"Features being used: {features}")
print(f"Total samples: {len(X)}")

# 2. Time Series Split (Sliding Window / Walk-Forward Validation)
tscv = TimeSeriesSplit(n_splits=5)

model = RandomForestRegressor(n_estimators=100, random_state=42)

mse_scores = []
rmse_scores = []
mape_scores = []

fold = 1
for train_index, test_index in tscv.split(X):
  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Train model
  model.fit(X_train, y_train)

  # Predict
  preds = model.predict(X_test)

  # Evaluate
  mse = mean_squared_error(y_test, preds)
  rmse = np.sqrt(mse)
  mape = mean_absolute_percentage_error(y_test, preds)

  mse_scores.append(mse)
  rmse_scores.append(rmse)
  mape_scores.append(mape)

  print(
      f"Fold {fold} -> RMSE: {rmse:.4f} | MAPE: {mape*100:.2f}% (Train size:"
      f" {len(X_train)}, Test size: {len(X_test)})"
  )
  fold += 1

print("\n--- Cross-Validation Summary ---")
print(f"Average RMSE: {np.mean(rmse_scores):.4f}")
print(f"Average MAPE: {np.mean(mape_scores)*100:.2f}%")
```

```python
import joblib
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_percentage_error, mean_squared_error
from sklearn.model_selection import TimeSeriesSplit
from statsmodels.tsa.stattools import adfuller
import yfinance as yf

# 1. Fetch Live Stock Data via yfinance
ticker_symbol = "AAPL"  # You can change this to any valid ticker symbol
print(f"Fetching live data for {ticker_symbol} from yfinance...")

# Download data from 2023-01-01 to today
df = yf.download(ticker_symbol, start="2023-01-01", end="2026-10-20")

# Clean up multi-index columns if returned by yfinance
if isinstance(df.columns, pd.MultiIndex):
  df.columns = df.columns.get_level_values(0)

# Ensure 'Close' column exists and drop missing values
df = df[["Close"]].dropna()
print(f"Data range: {df.index.min().date()} to {df.index.max().date()}")
print(f"Total records: {len(df)}")

# 2. Stationarity Check (ADF Test)
ts = df["Close"]


def test_stationarity(series):
  result = adfuller(series.dropna())
  print(f"ADF Statistic: {result[0]:.4f} | p-value: {result[1]:.4f}")


print("\n--- Original Series ADF Test ---")
test_stationarity(ts)

# 3. Feature Engineering: Lags & Rolling Means
data = pd.DataFrame({"Close": ts})

# Create 5 lag features
for i in range(1, 6):
  data[f"lag_{i}"] = data["Close"].shift(i)

# Create rolling mean features
data["rolling_mean_7"] = data["Close"].shift(1).rolling(window=7).mean()
data["rolling_mean_30"] = data["Close"].shift(1).rolling(window=30).mean()

# Drop rows with NaN values resulting from windows/lags
data = data.dropna()

# 4. Train/Test Split & Walk-Forward Validation
features = [col for col in data.columns if col != "Close"]
X = data[features]
y = data["Close"]

tscv = TimeSeriesSplit(n_splits=5)
model = RandomForestRegressor(n_estimators=100, random_state=42)

mape_scores = []
for train_index, test_index in tscv.split(X):
  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  model.fit(X_train, y_train)
  preds = model.predict(X_test)
  mape = mean_absolute_percentage_error(y_test, preds)
  mape_scores.append(mape)

print(f"\n--- Model Evaluation ---")
print(f"Average Walk-Forward MAPE: {np.mean(mape_scores)*100:.2f}%")

# 5. Train Final Model on All Recent Data and Save
final_model = RandomForestRegressor(n_estimators=100, random_state=42)
final_model.fit(X, y)

joblib.dump(final_model, "stock_rf_model.pkl")
print("✅ Updated model trained on recent data and saved as 'stock_rf_model.pkl'!")
```
